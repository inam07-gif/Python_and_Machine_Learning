#AUTOGRAD---> Automatic differentiation for tensor operations


```python
import torch
```

```python
x=torch.tensor(3.0, requires_grad=True)
```

```python
y=x**2
```

```python
print(x)
print(y)
```

```python
y.backward()
```

```python
x.grad
```

```python
x=torch.tensor(4.0, requires_grad=True)
```

```python
y=x ** 2
```

```python
z=torch.sin(y)
```

```python
x
```

```python
y
```

```python
z
```

```python
z.backward()
```
