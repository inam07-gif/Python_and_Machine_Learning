# 1.4 Remove Punctuation

````python
import string,time
````

````python
string.punctuation
````

````python
exclude=string.punctuation
````

````python
exclude
````

````python
def remove_punc(text):
    for char in exclude:
        text=text.replace(char,'')
    return text
````

````python
df['review']=df['review'].apply(remove_punc)
````

````python
df['review'][3]
````

````python
# or there is another way to remove punctuation, and it is way faster than the previous one
def remove_punc2(text):
    return text.translate(str.maketrans('','',exclude))
````

````python
df['review']=df['review'].apply(remove_punc2)
````

````python
slang_dict = {}

with open("sms_slang_translator/slang.txt", "r") as f:
    for line in f:
        key, value = line.strip().split("=", 1)
        slang_dict[key] = value
````

````python
slang_dict
````

````python
def chat_conversion(text):
    new_text=[]
    for w in text.split():
        if w.upper() in slang_dict:
            new_text.append(slang_dict[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
````

````python
chat_conversion('I WYWH')
````

# 1.7 Remove Stopwords
        

````python
from nltk.corpus import stopwords
````

````python
stopwords.words('english')
````

````python
stopwords.words('arabic')
````

````python
def remove_stop(text):
    new_txt=[]
    for w in text.split():
        if w in stopwords.words('english'):
            new_txt.append('')
        else:
            new_txt.append(w)
    x=new_txt[:]
    new_txt.clear()
    return " ".join(x)
    
        
````

# 4 Stemming:-> 
```Stemming in NLP means cutting a word down to its basic/root-like form by removing prefixes or suffixes.  ```

````python
# Stemming is mainly used is information retrival system(for example google search engine).
# For stemming we will use the PorterStemmer() from nltk library especially for english.
````

````python
from nltk.stem.porter import PorterStemmer
````

````python
sen3
````

````python
ps=PorterStemmer()
````

````python
def stem(text):
    return ' '.join([ps.stem(w) for w in text.split()])
````

````python
stem(sen3)
````

# 5 LEMMATIZATION

```LEMMATIZATION word → understand linguistic form → dictionary lemma```

````python
from nltk.stem import WordNetLemmatizer
````

````python
wl=WordNetLemmatizer()
````

````python
sen3
````
