```python
import numpy as np
import pandas as pd
import math
```

```python
import random
```

```python
x1=[random.uniform(1,100) for i in range(45)]

x2=[random.uniform(1,100) for i in range(45)]

y=random.choices([0,1],k=45)
```

```python
df=pd.DataFrame(
    {
        "x1":x1,
        "x2":x2,
        "y":y
    }
)
```

```python
df
```

```python
x=df[['x1','x2']]
```

```python
y=df['y']
```

```python
x
```

```python
w=[0,0]
b=0
```

```python
lr=0.001
```

```python
def sigmoid(z):
  return 1/(1+(2.7182)**-z)
```

```python
for j in range(1000):
  for i in range(len(x1)):
    x1c=x1[i]
    x2c=x2[i]
    z=np.dot(w,[x1c,x2c])+b
    y_pred=sigmoid(z)

    loss=(-1)*(y[i]*math.log(y_pred)+(1-y[i])*math.log(1-y_pred))
    error=y_pred-y[i]
    w[0]=w[0]-lr*(error)*x1c
    w[1]=w[1]-lr*(error)*x2c
    result=0
    b=b-lr*(error)
    if y_pred < 0.5:
      result=0
    elif y_pred >=0.5:
      result=1
  print(f" result:{result} | Loss:{loss} | w1:{w[0]} | w2:{w[1]} | pred: {y_pred} | error:{error}")



```
